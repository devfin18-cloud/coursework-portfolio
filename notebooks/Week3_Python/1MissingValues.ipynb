Missing Values and Basic Data Checks

The ff data will be used:
- `Alarm Survey Data.xlsx`
- `Student Grades.xlsx`

Kindly keep the data files in the same folder as this notebook and update the file when reading the data

```python
# lets start by importing the libraries we need
import pandas as pd
import numpy as np
```

```python
# load the alarm survey data and look at the first few rows.
alarm = pd.read_excel("AlarmSurveyData.xlsx") # set the directory based on where you stored the data on your computer
alarm.head()
```

```python
# check the structure of the data.
alarm.info()
```

```python
# check for missing values in each column.
alarm.isnull().sum()
```

```python
# look at the sleep quality column.
alarm["sleep_quality"].value_counts(dropna=False)
```

```python
# get the mean sleep quality for values that are 4.0 or above.
alarm[alarm["sleep_quality"] >= 4.0]["sleep_quality"].mean()
```

```python
# fill missing values in sleep quality with 4.5.
alarm["sleep_quality"] = alarm["sleep_quality"].fillna(4.5)
alarm.isnull().sum()
```

```python
# lets now load the student grades data.
grades = pd.read_excel("StudentGrades.xlsx") 
grades.head()
```

```python
grades.info()
```

```python
# look at the Year column values.
grades["Year"].value_counts(dropna=False)
```

```python
grades.isnull().sum()
```

```python
# mean of the Grade column.
grade_mean = grades["Grade"].mean()
grade_mean
```

```python
# let's fill missing Grade values with the mean grade.
grades["Grade"] = grades["Grade"].fillna(grade_mean)
grades.isnull().sum()
```

```python
# Look at only the rows where the Class is Freshman Seminar.
grades[grades["Class"] == "Freshman Seminar"]
```

```python
# update the Year column when the Class is Freshman Seminar.
grades["Year"] = np.where(grades["Class"] == "Freshman Seminar", "Freshman", grades["Year"])
grades["Year"].value_counts(dropna=False)
```

```python
# check the cleaned result.
grades.head()
```
