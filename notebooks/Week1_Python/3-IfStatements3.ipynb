# MBA 7100: If Statements


## What does `if` do?
`if` lets Python choose what to do based on a condition.

```python
sales = 1200

if sales >= 1000:
    print("Sales target reached!")
```

## if / else example

```python
profit = 150

if profit > 0:
    print("The business made a profit.")
else:
    print("The business did not make a profit.")
```

## if / elif / else example

```python
profit = 0

if profit > 0:
    print("Profit")
elif profit == 0:
    print("Break-even")
else:
    print("Loss")
```

## User input example
This checks whether a number is zero, positive, or negative.

```python
num = int(input("Enter a number: "))

if num == 0:
    print("Your number is 0")
elif num > 0:
    print("Your number is positive")
else:
    print("Your number is negative")
```

## Guess My Number logic
This shows the `if` logic used inside the game.

```python
secret_number = 42
guess = 30

if guess < secret_number:
    print("Your guess is too low")
else:
    print("Your guess is too high")
```

## Customer discount example

```python
purchase_amount = 120

if purchase_amount >= 100:
    print("Apply discount")
else:
    print("No discount")
```

## Key idea
Use `if` when you want different actions for different situations.
