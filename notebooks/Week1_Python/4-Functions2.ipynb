# MBA7100 Functions: No Parameters, No Return Value


## Notes
- A function lets us group reusable code.
- `def` means define a function.
- These first examples do not take input and do not return a value.

```python
# define a simple function
def greet():
    print('Hello, class!')
```

```python
# call the function
greet()
```

```python
# call the function more than once
greet()
greet()
```

```python
# define a function that prints a line
def print_line():
    print('--------------------')
```

```python
# use both functions together
print_line()
greet()
print_line()
```

```python
# define a business-style function
def show_course_name():
    print('MBA 7100: Business Intelligence, Machine Learning, and Predictive Analysis')
```

```python
# call the business-style function
show_course_name()
```

```python
# define a function that prints a short report header
def report_header():
    print('Weekly Sales Report')
    print('Store A')
```

```python
# call the report header function
report_header()
```

```python
# Task: write a function named say_welcome that prints Welcome!

def say_welcome():
    print('Welcome!')

say_welcome()
```

```python
# another short task

def show_break():
    print('Take a short break and come back.')

show_break()
```

```python
# reuse functions together
print_line()
show_course_name()
report_header()
print_line()
```
