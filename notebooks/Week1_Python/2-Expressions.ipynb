## 1. Arithmetic expressions

Python can do common arithmetic for us.

```
num1 = 5
num2 = 2

print("num1 + num2 =", num1 + num2)
print("num1 - num2 =", num1 - num2)
print("num1 * num2 =", num1 * num2)
print("num1 / num2 =", num1 / num2)
print("num1 // num2 =", num1 // num2)
print("num1 % num2 =", num1 % num2)
print("num1 ** num2 =", num1 ** num2)
```

**Note:**  
- `/` gives regular division  
- `//` gives integer division  
- `%` gives the remainder  
- `**` means "raise to a power"

## 2. A business example: revenue

A simple expression can represent a business calculation.

```
price = 12.5
quantity = 8
revenue = price * quantity

print("Price =", price)
print("Quantity =", quantity)
print("Revenue =", revenue)
```

## 3. Order of operations

Python follows the usual order of operations.
Parentheses help make expressions clear.

```
x = 1 + 3 * 5 / 2
print("x =", x)

discounted_revenue = (price * quantity) - 10
print("Discounted revenue =", discounted_revenue)
```

Try reading this carefully:

- multiplication and division happen before addition
- parentheses are a good way to make your intention clear

## 4. Implicit type conversion

Sometimes Python changes the result type for you.

```
a = 6
b = 2
c = 1.9

print("a + b =", a + b, type(a + b))
print("a / b =", a / b, type(a / b))
print("c + 1 =", c + 1, type(c + 1))
```

## 5. Explicit type conversion

We often need to convert values ourselves.
This is especially important when using `input()`.

```
unit_price_text = "19.99"
units_text = "12"

unit_price = float(unit_price_text)
units = int(units_text)

total_sales = unit_price * units

print("Unit price:", unit_price, type(unit_price))
print("Units:", units, type(units))
print("Total sales:", total_sales)
```

## 6. A quick input example

In VS Code notebooks, the input prompt may appear at the top of the screen.
Type the value and press **Enter**.

```
price = float(input("Enter unit price: "))
quantity = int(input("Enter quantity sold: "))

revenue = price * quantity
print("Revenue is:", revenue)
```

## 7. Floats are not always exact

This is normal in computing.

```
num3 = 1.9
print("num3 - 1 =", num3 - 1)
```

## 8. Shorthand assignment

These are short ways to update a variable.

```
customers = 10
print("Starting customers:", customers)

customers += 5
print("After increase:", customers)

customers -= 2
print("After decrease:", customers)

customers *= 3
print("After multiplication:", customers)
```

## 9. Average sales example

This is similar to the average example from class, but framed as a business example.

```
sales_week1 = float(input("Enter sales for week 1: "))
sales_week2 = float(input("Enter sales for week 2: "))

average_sales = (sales_week1 + sales_week2) / 2
print("Average sales =", average_sales)
```

## 10. Constants

Constants are values we give names to so the code is easier to read.
By convention, constants are written in capital letters.

```
TAX_RATE = 0.0625

price_before_tax = 100
tax_amount = price_before_tax * TAX_RATE
total_price = price_before_tax + tax_amount

print("Price before tax:", price_before_tax)
print("Tax amount:", tax_amount)
print("Total price:", total_price)
```
