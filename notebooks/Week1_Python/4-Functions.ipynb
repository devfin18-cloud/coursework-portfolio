# Functions in Python

## 1. A function with no parameters and no return value

```python
def print_intro():
    # this function prints a short welcome message.
    print("Welcome to MBA 7100")
    print("Today we are learning about functions.")

print_intro()
```

## 2. A function with one parameter
A parameter lets us pass information into a function.

```python
def print_opinion(score):
    # this prints a message based on the score.
    if score == 5:
        print("Excellent rating!")
    else:
        print("Thanks for the feedback.")

print_opinion(5)
print_opinion(3)
```

## 3. A function that returns a value
the return statement sends a value back to the caller.

```python
def meters_to_cm(meters):
    # convert meters to centimeters.
    return 100 * meters

result = meters_to_cm(5.2)
print("Centimeters:", result)
```

## 4. Return vs print
it can return a value to be stored or used later

```python
def meters_to_cm_case1(meters):
    # better for reuse: return the value.
    return 100 * meters

def meters_to_cm_case2(meters):
    # This only prints the value.
    print(100 * meters)

value = meters_to_cm_case1(3.5)
print("Stored value:", value)

meters_to_cm_case2(3.5)
```

## 5. A function with two parameters and a return value

```python
def average(a, b):
    # returns the average of two numbers.
    total = a + b
    return total / 2

mid = average(5.0, 10.2)
print("Average:", mid)
```

## 6. Multiple return statements
the function can be returned from different places.

```python
def max_value(num1, num2):
    # returns the larger number.
    if num1 >= num2:
        return num1
    return num2

larger = max_value(5, 1)
print("Larger number:", larger)

larger = max_value(1, 5)
print("Larger number:", larger)
```

## 7. A short business example
functions can be used to keep calculations clear and reusable.

```python
def revenue(price, quantity):
    # returns total revenue
    return price * quantity

sales = revenue(12.5, 8)
print("Revenue:", sales)
```

## 8. Quick recap
- Functions help us organise code.
- Parameters let us pass information in.
- `return` sends information back.
- Returning is not the same as printing.
