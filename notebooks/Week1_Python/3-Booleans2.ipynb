# MBA 7100: Booleans


## What is a boolean?
A boolean value is either `True` or `False`.

```python
is_profitable = True
is_closed = False

print(is_profitable)
print(is_closed)
```

## Comparison operators
These create boolean values.

```python
sales = 1200
cost = 900

print(sales > cost)
print(sales == cost)
print(cost != 900)
print(sales >= 1000)
```

## Boolean variables
We can store the result of a test in a variable.

```python
met_target = 850 >= 1000
is_break_even = 500 == 500

print("Met target?", met_target)
print("Break-even?", is_break_even)
```

## Logical operators
Use `and`, `or`, and `not` to combine boolean values.

```python
sales = 1200
cost = 800
has_inventory = True

print((sales > cost) and has_inventory)
print((sales < cost) or has_inventory)
print(not has_inventory)
```

## Range check example
This checks whether a score is between 70 and 100.

```python
score = 85
is_valid_score = (70 <= score) and (score <= 100)

print("Valid score?", is_valid_score)
```

## Key idea
Booleans help Python make decisions.
