## Notes
- A `for` loop repeats a block of code.
- `range(10)` gives numbers from `0` to `9`.
- We often use loops to repeat a calculation or print a pattern.

## Start with basic counting

```python
# print numbers 0 to 9
for i in range(10):
    print(i)
```

```python
# print numbers 1 to 5
for i in range(1, 6):
    print(i)
```

```python
# count by 2s
for i in range(0, 11, 2):
    print(i)
```

```python
# total sales over 5 days
sales = [120, 135, 128, 140, 150]
total_sales = 0
for sale in sales:
    total_sales = total_sales + sale
print('Total sales:', total_sales)
```

```python
# count the number of sales entries
sales = [120, 135, 128, 140, 150]
count = 0
for sale in sales:
    count = count + 1
print('Count:', count)
```

```python
# print a simple table of day and sales
sales = [120, 135, 128, 140, 150]
print('day', 'sales')
for i in range(len(sales)):
    print(i, sales[i])
```

```python
# nested for loop like the i, j example
print('i', 'j')
for i in range(2):
    for j in range(3):
        print(i, j)
```

```python
# create a small price and quantity table
prices = [10, 12, 15]
quantities = [2, 4]
print('price', 'quantity')
for price in prices:
    for quantity in quantities:
        print(price, quantity)
```

```python
# compute revenues for several products
prices = [10, 12, 15]
quantities = [3, 5, 2]
for i in range(len(prices)):
    revenue = prices[i] * quantities[i]
    print('Revenue for item', i, 'is', revenue)
```
