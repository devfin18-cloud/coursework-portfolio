## What is a while loop?
A `while` loop repeats **as long as a condition is True**.

```python
count = 1
while count <= 5:
    print("Count:", count)
    count = count + 1
```

## Weekly sales target example
We will keep adding weekly sales until the target is reached.

```python
target = 1000
sales_so_far = 0
week = 1

while sales_so_far < target:
    print("Week", week, "- sales so far:", sales_so_far)
    sales_so_far = sales_so_far + 250
    week = week + 1

print("Target reached:", sales_so_far)
```

## Sentinel loop example
A sentinel loop stops when the user enters a special value.

Here, `-1` means **stop**.

```python
total = 0
num = int(input("Enter a number (-1 to stop): "))

while num != -1:
    total = total + num
    num = int(input("Enter a number (-1 to stop): "))

print("Total is", total)
```

## Guess My Number example
This shows a `while` loop working with an `if` statement.

lets use a fixed secret number, you can change the number but it must be between the range

```python
secret_number = 42
print("I am thinking of a number between 1 and 99...")

guess = int(input("Enter a guess: "))

while guess != secret_number:
    if guess < secret_number:
        print("Your guess is too low")
    else:
        print("Your guess is too high")
    print("")
    guess = int(input("Enter a new guess: "))

print("Congrats! The number was:", secret_number)
```

## Key idea
- Start with a condition
- Update something inside the loop
- Make sure the loop can stop
