# MBA7100 Functions: Functions that Return Values


## Notes
- A return statement ends the function and gives a value back.
- We can store the returned value in a variable.

```python
# function that returns an average
def average(a, b):
    total = a + b
    return total / 2
```

```python
# store the result
mid = average(5.0, 10.2)
print(mid)
```

```python
# use the result inside print
print(average(20, 30))
```

```python
# function that returns revenue
def calculate_revenue(price, quantity):
    return price * quantity
```

```python
# call and store the result
revenue = calculate_revenue(15, 6)
print('Revenue:', revenue)
```

```python
# function that returns profit
def calculate_profit(revenue, cost):
    return revenue - cost
```

```python
# use one returned value inside another
revenue = calculate_revenue(20, 10)
profit = calculate_profit(revenue, 120)
print('Profit:', profit)
```

```python
# function that returns the larger of two numbers
def larger_value(a, b):
    if a > b:
        return a
    else:
        return b
```

```python
# test the larger_value function
print(larger_value(9, 14))
```

```python
# function with multiple possible returns
def number_sign(num):
    if num == 0:
        return 'zero'
    elif num > 0:
        return 'positive'
    else:
        return 'negative'
```

```python
# test number_sign
print(number_sign(12))
print(number_sign(-3))
print(number_sign(0))
```
