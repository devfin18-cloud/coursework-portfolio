## 3. Variables

A variable is a place to store information in a program.

Think of a variable as a **label** attached to a value.

```python
x = 10
print(x)

x = 5
print(x)

x = 5 + 7
print(x)
```

### Short note
When you write `x = 10`, Python stores the value `10` and gives it the label `x`.

When you assign a new value, the variable changes.

## 4. Variable names

Use names that help you remember what the value means.

```python
num_students = 10
num_in_class = 10
num_absent = num_students - num_in_class

print("Students:", num_students)
print("In class:", num_in_class)
print("Absent:", num_absent)
```

## 5. Types

Python values come in different types.
- `int` → whole numbers
- `float` → numbers with decimals
- `str` → text
- `bool` → `True` or `False`

```python
num_students = 10          # int
price = 19.99              # float
course_name = "MBA 7100"   # str
is_online = False          # bool

print(type(num_students))
print(type(price))
print(type(course_name))
print(type(is_online))
```

## 6. `input()` gives text back

When we use `input()`, Python gives us back a **string** (text).

So if the user enters a number, we often convert it with `int()` or `float()`.

# in VS Code or Jupyter notebook, the input might appear at the top of the window

```python
sample_text = "9"
print(sample_text)
print(type(sample_text))

sample_number = int(sample_text)
print(sample_number)
print(type(sample_number))
```

## 7. A simple business example

Here is a small MBA-style example.

We will ask for:
- unit price
- quantity sold

Then we will calculate revenue.

**Note:** This is the same basic idea as the addition example from class, but now the example is business-focused.

```python
print("This program calculates revenue.")
price = float(input("Enter unit price: "))
quantity = int(input("Enter quantity sold: "))
revenue = price * quantity
print("Revenue is:", revenue)
```

## 8. What's going on here?

Step by step:
1. `input()` collects text from the user
2. `float(price)` converts the price to a decimal number
3. `int(quantity)` converts the quantity to a whole number
4. `revenue = price * quantity` does the calculation
5. `print()` shows the result

## 9. Printing text and values together

You can print in different ways.

```python
revenue = 1250.75

print("Revenue is " + str(revenue))
print("Revenue is", revenue)
```

## 10. Multiple values in `print()`

When you separate items with commas, Python prints them with spaces.

```python
x = 4
y = 0.2

print("x =", x)
print("y =", y)
print("x =", x, "and y =", y)
```
