# MBA7100 Functions: Parameters, No Return Value


## Notes
- Parameters are the values a function expects.
- Arguments are the values we pass in when we call the function.

```python
# function with one parameter
def greet_person(name):
    print('Hello,', name)
```

```python
# call the function
greet_person('Mike')
```

```python
# call it again with a different value
greet_person('Tony')
```

```python
# function with two parameters
def show_revenue(price, quantity):
    revenue = price * quantity
    print('Revenue is', revenue)
```

```python
# call the revenue function
show_revenue(10, 5)
```

```python
# another call
show_revenue(12.5, 8)
```

```python
# function that prints a discount message
def show_discount(price, rate):
    discount = price * rate
    print('Discount amount is', discount)
```

```python
# call the discount function
show_discount(100, 0.10)
```

```python
# function with three parameters
def show_profit(revenue, cost, month):
    profit = revenue - cost
    print('Profit for', month, 'is', profit)
```

```python
# call the profit function
show_profit(5000, 3800, 'March')
```

```python
# Task: complete the function so it prints total cost
# Hint: total cost = price * quantity

def show_total_cost(price, quantity):
    total_cost = price * quantity
    print('Total cost is', total_cost)

show_total_cost(7, 4)
```

```python
# quick practice with a loop

def show_square(number):
    print(number, 'squared is', number * number)

for i in range(1, 6):
    show_square(i)
```
